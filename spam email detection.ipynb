```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
texts = []
labels = []

with open("spam.csv", "r", encoding="latin-1") as f:
    next(f)

    for line in f:
        line = line.rstrip("\r\n")

        if "," not in line:
            continue

        text, label = line.rsplit(",", 1)
        label = label.strip().lower()

        if label in ["spam", "ham"] and text.strip():
            texts.append(text.strip())
            labels.append(label)

print("CSV data loaded successfully!")
print("Total records:", len(texts))
```

```python
data = pd.DataFrame({
    "text": texts,
    "label": labels
})

print(data.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data["text"],
    data["label"],
    test_size=0.2,
    random_state=42,
    stratify=data["label"]
)

print("Training data:", len(X_train))
print("Testing data:", len(X_test))
```

```python
vectorizer = CountVectorizer()

X_train_vector = vectorizer.fit_transform(X_train)
X_test_vector = vectorizer.transform(X_test)

print("Text converted into numerical data!")
```

```python
model = MultinomialNB()

model.fit(X_train_vector, y_train)

print("Spam detection model trained successfully!")
```

```python
predictions = model.predict(X_test_vector)

accuracy = accuracy_score(y_test, predictions)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
email = input("Enter email text: ")

email_vector = vectorizer.transform([email])
result = model.predict(email_vector)[0]

if result == "spam":
    print("Result: SPAM EMAIL")
else:
    print("Result: NOT SPAM EMAIL")
```
